```python
#categorical encoding :

import pandas as pd

from sklearn.preprocessing import LabelEncoder

data = {
    'Gender': ['Male', 'Female', 'Female', 'Male', 'Female'],
    'City': ['Mumbai', 'Delhi', 'Chennai', 'Mumbai', 'Delhi'],
    'Size': ['Small', 'Large', 'Medium', 'Large', 'Small'],
    'Age': [25, 45, 32, 60, 28],
    'Fare': [15, 300, 85, 450, 20]
}

df = pd.DataFrame(data)
print(df)


```

```python
#label encoding of Gender column
le = LabelEncoder()
df['Gender'] = le.fit_transform(df['Gender'])

print(df)
```

```python
#one-hot encoding of city column , dropped one column
encoded_df = pd.get_dummies(df, columns=['City'], drop_first=True)

print(encoded_df)
```

```python
#label encoding of Size column
encoded_df['Size'] = df['Size'].map({'Small': 0, 'Medium': 1, 'Large': 2})

print(encoded_df)
```

##                                                                **SCALING**

```python
# standard scaling for Age column as it does not have any very high values or outliers...the values increases smoothly.

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

encoded_df['Age'] = scaler.fit_transform(df[['Age']])

stand_scaled_df = encoded_df

print(stand_scaled_df)
```

```python
# Robust Scaling:
from sklearn.preprocessing import RobustScaler

scaler = RobustScaler()

stand_scaled_df['Fare'] = scaler.fit_transform(stand_scaled_df[['Fare']])

rob_scaled_df = stand_scaled_df

preprocessed_data = rob_scaled_df

print(preprocessed_data)

type(preprocessed_data)
```
